# Automatic File Sorter in File Explorer

```python
import os, shutil
```

```python
# Getting the file path from the user
path = input(r"Enter the path of the file: ") # Enter "C:/Users/Shriyansh Singh/Pictures/Camera Roll/Phone Photos & Videos/Samsung S21/"
# Finding the content of the file to be sorted
file_name = os.listdir(path)
```

```python
import keyboard

# Creating a list of folder names to be created for sorting the files
folder_names = []
# Store the names of the folders entered by the user in a list, until user hits escape
while True:
    folder_name = input(f"Enter the name for the sorting folder (or press 'esc' to stop): ")
    folder_names.append(folder_name)

    if keyboard.is_pressed('esc'):
        break

# Delete the last entry which is 'esc'
folder_names.pop()
# Print out the folder names for confirmation
print(folder_names)
```

```python
# Generating folders for sorting the files with file names
for name in folder_names:
    if not os.path.exists(path + name):
        os.makedirs(path + name)

# Sorting the files into the respective folders
for file in file_name:
    jpg_folder = os.path.exists(path + 'jpgs/' + file)
    jpeg_folder = os.path.exists(path + 'jpegs/' + file)
    heic_folder = os.path.exists(path + 'heics/' + file)
    dng_folder = os.path.exists(path + 'dngs/' + file)
    png_folder = os.path.exists(path + 'pngs/' + file)
    webp_folder = os.path.exists(path + 'webps/' + file)
    mp4_folder = os.path.exists(path + 'mp4s/' + file)

    if '.jpg' in file and not jpg_folder:
        shutil.move(path + file, path + 'jpgs/' + file)
    elif '.jpeg' in file and not jpeg_folder:
        shutil.move(path + file, path + 'jpegs/' + file)
    elif '.heic' in file and not heic_folder:
        shutil.move(path + file, path + 'heics/' + file)
    elif '.dng' in file and not dng_folder:
        shutil.move(path + file, path + 'dngs/' + file)
    elif '.mp4' in file and not mp4_folder:
        shutil.move(path + file, path + 'mp4s/' + file)
    elif '.png' in file and not png_folder:
        shutil.move(path + file, path + 'pngs/' + file)
    elif '.webp' in file and not webp_folder:
        shutil.move(path + file, path + 'webps/' + file)
```
