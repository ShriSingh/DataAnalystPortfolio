## Amazon Web Scraper

Getting the libraries

```python
# import libraries
from bs4 import BeautifulSoup
import requests
import time
import datetime

import smtplib
```

Connecting to the website selected

```python
# Getting the url link from the user
# Sample URLs: 
# https://www.amazon.com/Fosi-Audio-Headphone-Converter-Microphone/dp/B0DZCTM9N6/ref=pd_ci_mcx_mh_mcx_views_0_title
# https://www.amazon.com/dp/B0C8ZJKPWC/?coliid=I11I88XZGUAG4N&colid=30PDXW24ZK6DN&ref_=list_c_wl_lv_ov_lig_dp_it
URL = input("Enter the URL: ")
# Getting the user agent and other values using https://httpbin.org/get
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36", "Accept-Encoding": "gzip, deflate, br, zstd", "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8", "DNT": "1", "Connection": "close", "Upgrade-Insecure-Requests": "1"}

# Loading up the url and the respective the user agent
page = requests.get(URL, headers=headers)

# Pulling in the content from the page
soup1 = BeautifulSoup(page.content, "html.parser")
soup2 = BeautifulSoup(soup1.prettify(), "html.parser") # Making HTMl more readable

# Getting the title of the product
title = soup2.find(id='productTitle').get_text()
# Getting the price of the product
price = soup2.find(id='corePriceDisplay_desktop_feature_div').get_text()
# Getting the rating of the product
# rating = soup2.find(id='averageCustomerReviews').get_text()

# print(Soup1)
# print(soup2)
print(title)
print(price)
# print(rating)
```

Formatting the title and price of the product data

```python
title = title.strip()[:31]
price = price.strip()[:7]

print(title)
print(price)
```

```python
import datetime

today = datetime.date.today()
print(today)
```

Getting the product data into a csv file

```python
import csv

# Setting up the header and the data in the list
header = ['Product', 'Price', 'Date']
data = [title, price, today]

with open('AmazonWebScraperDataset.csv', 'w', newline='', encoding='UTF8') as f:
    writer = csv.writer(f)
    writer.writerow(header)
    writer.writerow(data)
```
