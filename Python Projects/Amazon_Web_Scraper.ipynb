Getting the libraries

```python
# import libraries
from bs4 import BeautifulSoup
import requests
import time
import datetime

import smtplib
```

Connecting to the website selected

```python
# Getting the url link from the user
# Sample URL: https://www.amazon.com/Fosi-Audio-Headphone-Converter-Microphone/dp/B0DZCTM9N6/ref=pd_ci_mcx_mh_mcx_views_0_title
URL = input("Enter the URL: ")
# Getting the user agent and other values using https://httpbin.org/get
headers = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/137.0.0.0 Safari/537.36", "Accept-Encoding": "gzip, deflate, br, zstd", "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8", "DNT": "1", "Connection": "close", "Upgrade-Insecure-Requests": "1"}

# 
page = requests.get(URL, headers=headers)

# Pulling in the content from the page
Soup1 = BeautifulSoup(page.content, "html.parser")

print(Soup1)
```
