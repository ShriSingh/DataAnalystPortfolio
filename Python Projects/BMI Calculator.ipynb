# BMI Calculator

Body Mass Index(BMI) is a measurement of body fat based on height and weight.

```BMI = (weight in pounds * 703) / (height in inches)^2```

| BMI | Classification | Health Risk |
| ------ | -------------- | ----------- |
| < 18.5 | Underweight | Minimal |
| 18.5 - 24.9 | Normal Weight | Minimal |
| 25 - 29.9 | Overweight | Increased |
| 30 - 34.9 | Obese | High |
| 35 - 39.9 | Severely Obese | Very High |
| 40 & over | Morbidly Obese | Extremely High |


````python
# Noting down the weight of the user
weight = float(input("Please enter your weight: "))
weight_type_check = input("Is it in pounds or kilos?: ")
# Noting down the height of the user
feet_height = int(input("Please enter the \"feet\" of your height: "))
inches_height = int(input("Please enter the \"inches\" part of your height: "))

# kgs to lbs conversion
if weight_type_check == 'lbs' or 'lb' or 'Lbs' or 'Pounds' or 'pounds':
    pass
elif weight_type_check == 'kgs' or 'kg' or 'Kg' or 'Kilos' or 'kilos':
    weight *= 2.20462262

# Totaling the height in inches
height = (feet_height * 12) + inches_height

# Calculating BMI using the weight
calculated_bmi = (weight * 703) / (height ** 2)

# Printing out the BMI
print("Here's the calculated BMI:", calculated_bmi)

# Figuring out the health condition using BMI
if calculated_bmi < 18.5:
    message = "You are \"Underweight\" and have \"Minimal\" health risk..."
elif 18.5 <= calculated_bmi <= 24.9:
    message = "You have \"Normal Weight\" and have \"Minimal\" health risk..."
elif 25 <= calculated_bmi <= 29.9:
    message = "You are \"Overweight\" and have \"Increased\" health risk..."
elif 30 <= calculated_bmi <= 34.9:
    message = "You are \"Obese\" and have \"High\" health risk!"
elif 35 <= calculated_bmi <= 39.9:
    message = "You are \"Severely Obese\" and have \"Very High\" health risk!!"
else:
    message = "You are \"Morbidly Obese\" and have \"Extremely High\" health risk!!!"

print(message)
````
